### Short answers to comprehension questions (simple)
1. Why check data first? — To find bad rows, missing numbers, and understand the shape before modeling.
2. What does a missing value mean? — A number is not available for that year and country.
3. Why transpose? — So each row is a year and it is easier to plot time trends.
4. What does a histogram show? — How values are spread (many small values or few big ones).
5. What is a boolean mask? — A simple yes/no filter for rows (for example, years before 1990).
6. What does correlation tell us? — If two countries’ numbers move together; it does not prove cause.
7. Which model fit better? — Polynomial degree 2 fit better for Belgium (based on RMSE and R²).
8. Why can a model that fits old data still fail later? — The future may change in ways not in the old data.
9. What changed with a train/test split? — We measure how well the model works on new years it did not see during training.
10. One limit of using year alone — Other important factors (economy events) are not included.

**One thing I understand better now:** Check data first and split train/test to avoid overconfidence.
